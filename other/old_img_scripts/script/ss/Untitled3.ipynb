```python
import rasterio

# Define the path to the TIFF file
tiff_file_path = 'data/biomassters/train_zip/EU_TreeMap/AbiesSpp.tif'
with rasterio.open(tiff_file_path) as src:
        print("TIFF file read successfully.")
        # Read the first band of the TIFF file
        band1 = src.read(1)
        print(src.profile)
        print(src.count)
import matplotlib.pyplot as plt

# Visualize the first band of the TIFF file
plt.figure(figsize=(10, 10))
plt.imshow(band1, cmap='gray')
plt.title('Visualization of the First Band of the TIFF File')
plt.colorbar()
plt.show()

```

```python
import os

# Define the paths to the files
rrd_file_path = 'data/biomassters/train_zip/EU_TreeMap/AbiesSpp.rrd'
aux_file_path = 'data/biomassters/train_zip/EU_TreeMap/AlnusSpp.aux'

# Function to check if a file exists and print its size
def check_file(file_path):
    if os.path.exists(file_path):
        file_size = os.path.getsize(file_path)
        print(f"File '{file_path}' exists with size: {file_size} bytes.")
    else:
        print(f"File '{file_path}' does not exist.")

# Check the RRD and AUX files
check_file(rrd_file_path)
check_file(aux_file_path)
```
